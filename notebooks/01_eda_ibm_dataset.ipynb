```python
df2 = pd.read_csv('../data/raw/dataset.csv')
print(df2.shape)
print(df2.dtypes)
df2.head()
```

```python
print(df2.shape)
print(df2.dtypes)
print()
print("Missing values per column:")
print(df2.isnull().sum())
print()
print("Currency breakdown:")
print(df2['invoice_currency'].value_counts())
print()
print("isOpen breakdown:")
print(df2['isOpen'].value_counts())
```

```python
df2['invoice_currency'].value_counts()
```

```python
# Filter to USD only, drop fully-empty column, keep only settled invoices
df2_clean = df2[df2['invoice_currency'] == 'USD'].copy()
df2_clean = df2_clean.drop(columns=['area_business'])
df2_clean = df2_clean[df2_clean['isOpen'] == 0].copy()

# Convert date columns to actual dates
df2_clean['clear_date'] = pd.to_datetime(df2_clean['clear_date'])
df2_clean['due_in_date'] = pd.to_datetime(df2_clean['due_in_date'], format='%Y%m%d')

# Compute payment delay in days
df2_clean['days_late'] = (df2_clean['clear_date'] - df2_clean['due_in_date']).dt.days

print(df2_clean.shape)
print(df2_clean['days_late'].describe())
```

```python
df2_clean['days_late'].hist(bins=50)
plt.xlabel('Days Late (negative = early)')
plt.ylabel('Number of Invoices')
plt.title('Distribution of Payment Delay (SAP dataset)')
plt.show()

# What fraction pay early, on-time, late?
early = (df2_clean['days_late'] < 0).sum()
ontime = (df2_clean['days_late'] == 0).sum()
late = (df2_clean['days_late'] > 0).sum()
print(f"Early: {early} ({early/len(df2_clean)*100:.1f}%)")
print(f"On-time: {ontime} ({ontime/len(df2_clean)*100:.1f}%)")
print(f"Late: {late} ({late/len(df2_clean)*100:.1f}%)")
```

```python
customer_stats = df2_clean.groupby('cust_number')['days_late'].agg(['mean', 'std', 'count'])
customer_stats = customer_stats[customer_stats['count'] >= 5]  # only customers with enough invoices to be meaningful
print(customer_stats.sort_values('mean').head(10))
print()
print(customer_stats.sort_values('mean').tail(10))
```
